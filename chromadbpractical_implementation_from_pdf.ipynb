```python
pdf_files = ["resource 1.pdf","resource 2.pdf","resource 3.pdf","resource 4.pdf","resource 5.pdf","resource 6.pdf","resource 7.pdf"]
```

```python
import pymupdf 
final_chunk=[]
final_meta=[]
count=0
for resource_pdf in pdf_files:
    print(resource_pdf)
    doc=pymupdf.open(resource_pdf)
    count+=1
    metadata="source"
    for page in range(len(doc)):
        pagee=doc[page]
        text_page=pagee.get_text()
        text_page = text_page.replace("\n\n", "<<PARA>>")
        text_page = text_page.replace("\n", " ")
        text_page = text_page.replace("<<PARA>>", "\n\n")
        
        
        import math
        for j in range(math.ceil(len(text_page)/500)):
            text=text_page[j*500:(j+1)*500+100]
            final_chunk.append(text) 
            final_meta.append({metadata:"paper"+str(count)})
        # for text_splittedby_100 in split_text_100:
        #     print()
print(count)        
# print((final_chunk))  
print(len(final_chunk))

        
    
```

```python
from chromadb.utils.embedding_functions import OpenAIEmbeddingFunction
dict.update
import chromadb
client = chromadb.PersistentClient(path="./my_chromadb")
collection=client.create_collection(name="research_papers")
ids=[]
for i in range(len(final_chunk)):
    ids.append("id"+str(i+1))
collection.add(
    ids=ids,
    documents=final_chunk,
    metadatas=final_meta,
)
```
